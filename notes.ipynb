### Training

Pre-training : (Traditional VLA training) : Regular next-token Imitation learning with all components hot
Post-training : ActVLP paradigm::
```
            1) Step 1 : World knowledge post-training, frozen ViT, hot LLM, training on regular world knowledge text question with image context tokens from Vit
            2) Step 2 : Visual knowledge (based on the Vit's embeddings) questions with both Vit and llm hot
            3) Step 3 : Next token prediction imitation learning again as in pre-training, but with the Vit frozen this time
```
